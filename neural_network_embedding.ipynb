```python
import gzip
import shutil

gz_file = "GoogleNews-vectors-negative300.bin.gz"
output_file = "GoogleNews-vectors-negative300.bin"

with gzip.open(gz_file, 'rb') as f_in:
    with open(output_file, 'wb') as f_out:
        shutil.copyfileobj(f_in, f_out)

print("File extracted successfully!")
```

```python
from gensim.models import KeyedVectors

# Load pre-trained Word2Vec model (Google News)
model = KeyedVectors.load_word2vec_format("GoogleNews-vectors-negative300.bin", binary=True)

# Perform vector arithmetic
result = model.most_similar(positive=['king', 'woman'], negative=['man'], topn=1)
print(result)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Example word vectors
king = np.array([0.95, 0.96])
queen = np.array([0.96, 0.95])

# Compute cosine similarity
similarity = cosine_similarity([king], [queen])
print(similarity)
```

```python
from gensim.models import Word2Vec

# Sample dataset
sentences = [["the", "cat", "sits", "on", "the", "mat"],
             ["the", "dog", "plays", "in", "the", "yard"]]

# Training a CBOW model
model = Word2Vec(sentences, vector_size=50, window=3, min_count=1, sg=0)  # sg=0 means CBOW

# Get word embedding
print(model.wv["cat"])

```
